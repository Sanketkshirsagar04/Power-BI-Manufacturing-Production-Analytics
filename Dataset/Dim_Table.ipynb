**Dim Date**

```python
import pandas as pd
import calendar

# ==========================================
# Manufacturing Analytics
# Dim_Date Generator
# ==========================================

start_date = "2023-01-01"
end_date = "2025-12-31"

dates = pd.date_range(start=start_date, end=end_date, freq="D")

data = []

for d in dates:

    year = d.year
    month = d.month
    quarter = (month - 1) // 3 + 1

    data.append({

        "DateKey": int(d.strftime("%Y%m%d")),

        "Date": d.strftime("%Y-%m-%d"),

        "Year": year,

        "Quarter": quarter,

        "QuarterName": f"Q{quarter}",

        "Month": month,

        "MonthName": d.strftime("%B"),

        "MonthShort": d.strftime("%b"),

        "WeekOfYear": int(d.isocalendar().week),

        "Day": d.day,

        "DayName": d.strftime("%A"),

        "DayShort": d.strftime("%a"),

        "DayOfWeek": d.isoweekday(),

        "IsWeekend": 1 if d.weekday() >= 5 else 0,

        "IsBusinessDay": 0 if d.weekday() >= 5 else 1,

        "IsMonthStart": 1 if d.is_month_start else 0,

        "IsMonthEnd": 1 if d.is_month_end else 0,

        "IsQuarterStart": 1 if d.month in [1,4,7,10] and d.day == 1 else 0,

        "IsQuarterEnd": 1 if d.is_month_end and d.month in [3,6,9,12] else 0,

        "IsYearStart": 1 if d.month == 1 and d.day == 1 else 0,

        "IsYearEnd": 1 if d.month == 12 and d.day == 31 else 0,

        "FinancialYear": f"FY{year}",

        "MonthYear": d.strftime("%b-%Y"),

        "YearMonth": d.strftime("%Y-%m")

    })

df = pd.DataFrame(data)

# Save CSV
df.to_csv("Dim_Date.csv", index=False)

print("=" * 50)
print("Dim_Date.csv Created Successfully")
print(f"Rows : {len(df)}")
print(f"Columns : {len(df.columns)}")
print("=" * 50)

print(df.head())
```

```python
import pandas as pd
import random

# ==========================================================
# Manufacturing Analytics
# ViewSonic - Dim_Employee Generator
# ==========================================================

random.seed(42)

first_names = [
"James","John","Robert","Michael","William","David","Richard","Joseph",
"Thomas","Charles","Daniel","Matthew","Anthony","Mark","Donald","Steven",
"Paul","Andrew","Joshua","Kevin","Brian","George","Edward","Jason","Ryan",
"Jacob","Nathan","Aaron","Adam","Justin","Brandon","Tyler","Austin",
"Zachary","Christian","Patrick","Samuel","Benjamin","Noah","Logan",
"Emily","Olivia","Emma","Sophia","Isabella","Mia","Charlotte","Amelia",
"Harper","Evelyn","Abigail","Ella","Scarlett","Grace","Lily","Chloe"
]

last_names = [
"Smith","Johnson","Williams","Brown","Jones","Garcia","Miller","Davis",
"Rodriguez","Martinez","Hernandez","Lopez","Gonzalez","Wilson","Anderson",
"Thomas","Taylor","Moore","Jackson","Martin","Lee","Perez","Thompson",
"White","Harris","Sanchez","Clark","Ramirez","Lewis","Robinson","Walker",
"Young","Allen","King","Wright","Scott","Green","Baker","Adams","Nelson"
]

departments = [
"Production",
"Production",
"Production",
"Production",
"Quality",
"Maintenance",
"Warehouse",
"Procurement",
"Engineering"
]

job_titles = {
"Production":["Operator I","Operator II","Production Supervisor"],
"Quality":["Quality Inspector","Quality Engineer"],
"Maintenance":["Maintenance Technician","Maintenance Engineer"],
"Warehouse":["Warehouse Associate","Warehouse Supervisor"],
"Procurement":["Procurement Executive","Procurement Manager"],
"Engineering":["Manufacturing Engineer","Process Engineer"]
}

education = [
"High School",
"Diploma",
"Bachelor",
"Master"
]

employment = [
"Full-Time",
"Full-Time",
"Full-Time",
"Contract"
]

employees=[]

for i in range(1,201):

    dept=random.choice(departments)

    employees.append([

        i,

        f"EMP-{i:04}",

        random.choice(first_names)+" "+random.choice(last_names),

        random.choice(["Male","Female"]),

        random.randint(22,60),

        dept,

        random.choice(job_titles[dept]),

        random.randint(1,8),

        random.randint(1,25),

        random.choice(education),

        random.choice(employment),

        random.randint(42000,125000),

        random.randint(2008,2025),

        "Active"

    ])

columns=[

"EmployeeKey",
"EmployeeCode",
"EmployeeName",
"Gender",
"Age",
"Department",
"JobTitle",
"PlantKey",
"ExperienceYears",
"Education",
"EmploymentType",
"SalaryUSD",
"HireYear",
"Status"

]

df=pd.DataFrame(employees,columns=columns)

df.to_csv("Dim_Employee.csv",index=False)

print("="*60)
print("Dim_Employee.csv Created Successfully")
print("="*60)
print(df.head())
print("\nTotal Employees :",len(df))
```
